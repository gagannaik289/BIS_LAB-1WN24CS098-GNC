```python

import random

# Parameters
n = 5
Pa = 0.25
MaxIter = 100

subjects = ["DSA", "DBMS", "OS", "Java"]
slots = 4

# Generate random timetable
def generate_timetable():
    return random.sample(subjects, len(subjects))

# Calculate fitness
def fitness(timetable):
    clashes = 0

    for i in range(len(timetable)):
        for j in range(i + 1, len(timetable)):
            if timetable[i] == timetable[j]:
                clashes += 1

    return clashes

# Initialize nests
nests = [generate_timetable() for _ in range(n)]

best = min(nests, key=fitness)

# Cuckoo Search
for iteration in range(MaxIter):

    for i in range(n):

        # Generate new solution using random changes
        new = nests[i].copy()

        a, b = random.sample(range(slots), 2)
        new[a], new[b] = new[b], new[a]

        # Compare fitness
        j = random.randrange(n)

        if fitness(new) < fitness(nests[j]):
            nests[j] = new

    # Abandon worst nests
    for i in range(n):
        if random.random() < Pa:
            nests[i] = generate_timetable()

    # Update best solution
    current = min(nests, key=fitness)

    if fitness(current) < fitness(best):
        best = current.copy()

# Display result
print("Best Timetable:")

for i in range(slots):
    print("Slot", i + 1, ":", best[i])

print("Fitness:", fitness(best))
```
